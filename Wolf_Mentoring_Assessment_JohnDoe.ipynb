### Question 1: What do you understand about application re-writing and application modernization?

**Application re-writing** means rebuilding an existing application from scratch using a new technology stack, architecture, or design paradigm — essentially replacing the old system while preserving its business functionality. It is a high-risk, high-cost, "big bang" approach typically taken when the existing codebase is so technically compromised that incremental improvement is no longer viable.

**Application modernization** is a broader, more strategic term that describes the process of evolving legacy applications to meet current business, technical, and operational needs. It does not necessarily mean rewriting; it encompasses a spectrum of approaches:

| Modernization Strategy | Description | Example |
|------------------------|-------------|----------|
| **Rehost (Lift & Shift)** | Move the application to a new environment (e.g., cloud) with minimal code changes | Moving an on-premise Java app to AWS EC2 without refactoring |
| **Replatform** | Make targeted optimisations to leverage a new platform without changing core architecture | Moving from on-premise DB to Amazon RDS |
| **Refactor / Re-architect** | Restructure the internal code to improve performance, scalability, or maintainability without changing external behaviour | Decomposing a monolith into microservices incrementally |
| **Rebuild / Rewrite** | Discard the old system and rebuild from scratch with modern technology | Building a new event-driven microservices platform to replace a legacy monolith |
| **Replace** | Retire the custom application and adopt a commercial off-the-shelf (COTS) or SaaS solution | Replacing a custom CRM with Salesforce |

**Relevance to John's experience:** At Spireon, John led a practical application of modernization — transitioning from a monolithic telematics platform to a microservices architecture on AWS using Docker and ECS. This is a classic **re-architect** modernization strategy. The outcome — zero-downtime deployments, improved scalability, and 300+ new dealerships onboarded — illustrates why modernization (rather than a full rewrite) is often preferred: it de-risks the transition by keeping the system operational while incrementally improving it.

**Key distinction to remember:** Rewriting replaces everything at once (high risk); modernization is typically an iterative, phased journey (lower risk, more manageable). Best practice is to avoid the full rewrite unless absolutely necessary — the "strangler fig" pattern (incrementally replacing functionality in a live system) is a widely-used modernization approach that eliminates the need for a complete rewrite.

### Question 2: Can you explain the differences between Infrastructure as a Service (IaaS), Platform as a Service (PaaS), and Software as a Service (SaaS)?

These three cloud service models differ in how much of the technology stack the **cloud provider manages** versus how much the **customer manages**.

```
What YOU manage vs. what the PROVIDER manages:

┌─────────────────┬──────────────────┬──────────────────┬──────────────────┐
│ Stack Layer     │ On-Premises      │ IaaS             │ PaaS             │ SaaS
├─────────────────┼──────────────────┼──────────────────┼──────────────────┤
│ Applications    │ You              │ You              │ You              │ Provider
│ Data            │ You              │ You              │ You              │ Provider
│ Runtime         │ You              │ You              │ Provider         │ Provider
│ Middleware      │ You              │ You              │ Provider         │ Provider
│ OS              │ You              │ You              │ Provider         │ Provider
│ Virtualisation  │ You              │ Provider         │ Provider         │ Provider
│ Servers         │ You              │ Provider         │ Provider         │ Provider
│ Storage         │ You              │ Provider         │ Provider         │ Provider
│ Networking      │ You              │ Provider         │ Provider         │ Provider
└─────────────────┴──────────────────┴──────────────────┴──────────────────┘
```

**Infrastructure as a Service (IaaS)**

The provider delivers virtualised computing resources — servers, storage, and networking — over the internet. The customer manages everything above the hypervisor: operating system, middleware, runtime, applications, and data.

- **Control:** Maximum control over the stack
- **Responsibility:** Customer manages OS patches, security, scaling
- **Best for:** Teams that need full control or have unusual requirements
- **Examples:** AWS EC2, Microsoft Azure Virtual Machines, Google Compute Engine
- **John's context:** At Spireon, running Docker containers on ECS is an IaaS-adjacent model — the team provisioned and managed compute infrastructure on AWS.

---

**Platform as a Service (PaaS)**

The provider delivers a complete managed environment — OS, runtime, middleware, and infrastructure — so the customer only needs to deploy their application code and data.

- **Control:** Moderate — customer owns the application logic; provider handles the platform
- **Responsibility:** Customer manages code, configuration, and data
- **Best for:** Development teams that want to focus on building, not managing infrastructure
- **Examples:** AWS Elastic Beanstalk, Heroku, Google App Engine, Azure App Service
- **John's context:** Deploying the LTK SaaS product on a managed cloud environment would be a PaaS model — the engineering team writes the application; the platform handles scaling and runtime.

---

**Software as a Service (SaaS)**

The provider delivers a fully managed, ready-to-use application over the internet. The customer uses the software via a browser or API — they own none of the underlying infrastructure, platform, or application code.

- **Control:** Minimal — customer configures, not customises
- **Responsibility:** Provider handles everything; customer manages data and user access
- **Best for:** Business users who need a tool, not a platform
- **Examples:** Salesforce, HubSpot, Stripe, DataDog, GitHub, Slack
- **John's context:** John *built* SaaS products (the LTK ad marketplace) and also *consumed* SaaS tools (HubSpot after migrating from NetSuite; Stripe for payments; DataDog for observability).

---

**Simple analogy — Pizza as a Service:**

| Model | Analogy |
|-------|----------|
| On-premises | You make pizza at home from scratch |
| IaaS | You use a rented kitchen with ovens; you bring ingredients and cook |
| PaaS | You use a rented kitchen with everything prepared; you just assemble and bake |
| SaaS | You order from a restaurant; someone else makes, bakes, and delivers |

### Question 3: Can you explain the roles of a Scrum Master, Product Owner, and Development Team within the Scrum framework?

Scrum is an Agile framework that organises work into short, fixed-length iterations called **sprints** (typically 2 weeks). It defines three core roles, each with distinct accountabilities:

---

**Scrum Master**

The Scrum Master is a **servant-leader and process guardian**. They do not manage the team in a traditional sense; instead, they protect the team from external interruptions, facilitate Scrum ceremonies, and continuously improve how the team works.

| Responsibility | What This Looks Like in Practice |
|----------------|-----------------------------------|
| Facilitate ceremonies | Runs sprint planning, daily standups, retrospectives, and sprint reviews |
| Remove impediments | Unblocks engineers from dependencies, process friction, or cross-team issues |
| Coach the team | Educates the team on Scrum values and continuous improvement |
| Shield the team | Protects the sprint from scope creep or unplanned interruptions |
| Serve the organisation | Helps the broader organisation adopt Agile practices |

**Key point:** The Scrum Master does not assign tasks or make product decisions. They are accountable for the *process*, not the *product*.

**John's context:** When John "rescued a struggling project by redefining the MVP, prioritising critical stories, and setting realistic deadlines using agile methodology," he was operating with Scrum Master-level process discipline — even in a Director role.

---

**Product Owner (PO)**

The Product Owner is the **voice of the customer and the business**. They own the Product Backlog — the prioritised list of everything the product needs — and are accountable for maximising the value of the work the Development Team produces.

| Responsibility | What This Looks Like in Practice |
|----------------|-----------------------------------|
| Own the Product Backlog | Creates, refines, and prioritises user stories and acceptance criteria |
| Define the product vision | Communicates "why" behind every feature; aligns backlog to business goals |
| Represent stakeholders | Translates business, customer, and market needs into actionable backlog items |
| Accept or reject work | Decides whether sprint output meets the Definition of Done |
| Make trade-off decisions | Decides what gets built, what gets deferred, and what gets cut |

**Key point:** The PO makes decisions about *what* to build and *in what order*. They do not dictate *how* the team builds it.

**John's context:** John defined the 2–3-year product vision at LTK, used OKRs to convey engineering focus areas, and implemented Definition of Ready (DoR) for stories — all PO-adjacent behaviours that demonstrate he understands and can partner deeply with the Product Owner role.

---

**Development Team**

The Development Team (now called "Developers" in the 2020 Scrum Guide) is a **self-organising, cross-functional group** of professionals who do the work of creating a potentially shippable product increment every sprint.

| Responsibility | What This Looks Like in Practice |
|----------------|-----------------------------------|
| Self-organise | Decide *how* to implement backlog items without being told by a manager |
| Cross-functional | Members collectively possess all skills needed: design, development, testing, deployment |
| Deliver the increment | Produce a working, tested, potentially shippable product at the end of every sprint |
| Estimate effort | Provide sprint capacity estimates (story points or hours) to help the PO plan |
| Uphold quality | Own the Definition of Done; are collectively responsible for quality, not just QA |

**Key point:** The Development Team is accountable for *how* the work gets done and *ensuring it meets quality standards*.

---

**Summary Table:**

| Role | Accountable for | Answers the question |
|------|----------------|----------------------|
| Product Owner | Product value and backlog | *What* to build and *why*? |
| Scrum Master | Process and team effectiveness | *How* does the team work best? |
| Development Team | The increment (working software) | *How* to build it? |

### Question 4: Can you describe what a microservices architecture is and its advantages over a monolithic architecture?

**Monolithic Architecture**

A monolithic application is built as a **single, tightly coupled unit**. All components — user interface, business logic, data access layer, and integrations — are packaged and deployed together. A change to any single component requires redeploying the entire application.

```
MONOLITH
┌──────────────────────────────────────────────┐
│          Single Deployable Unit              │
│  ┌──────────┐ ┌──────────┐ ┌──────────────┐ │
│  │   UI     │ │ Business │ │  Data Access │ │
│  │  Layer   │ │  Logic   │ │    Layer     │ │
│  └──────────┘ └──────────┘ └──────────────┘ │
│  ┌──────────────────────────────────────┐    │
│  │        Single Database               │    │
│  └──────────────────────────────────────┘    │
└──────────────────────────────────────────────┘
         ↕ Deploy everything together
```

**Microservices Architecture**

A microservices application is decomposed into a collection of **small, independently deployable services**, each responsible for a specific business capability. Services communicate with each other via APIs (REST, gRPC) or message queues (Kafka, SQS).

```
MICROSERVICES
┌──────────────┐  ┌──────────────┐  ┌──────────────┐  ┌──────────────┐
│  User Auth   │  │  Payments    │  │  Analytics   │  │  Creator     │
│  Service     │  │  Service     │  │  Service     │  │  Profiles    │
│  (Node.js)   │  │  (Java)      │  │  (Python)    │  │  (Golang)    │
│  Own DB      │  │  Own DB      │  │  Own DB      │  │  Own DB      │
└──────┬───────┘  └──────┬───────┘  └──────┬───────┘  └──────┬───────┘
       └──────────────────┴──────────────────┴──────────────────┘
                          API Gateway / Message Bus
```

**Advantages of Microservices over Monolith:**

| Advantage | Explanation | Real-world example |
|-----------|-------------|--------------------|
| **Independent deployability** | Each service can be deployed, scaled, and updated without touching others | Fix a bug in the Payments service without redeploying the entire platform |
| **Technology flexibility** | Different services can use different languages, databases, and frameworks | Auth service in Node.js; analytics pipeline in Python; real-time service in Golang |
| **Scalability** | Scale only the services under load, not the entire application | Scale the Creator Analytics service during peak reporting hours, not the entire system |
| **Fault isolation** | A failure in one service doesn't bring down the entire application | A crash in the Recommendations service doesn't take down Checkout |
| **Team autonomy** | Each squad owns and operates its own service independently, enabling parallel development | Three squads can ship independently without waiting for each other's code |
| **Easier CI/CD** | Smaller codebases mean faster build, test, and deploy cycles | A single microservice can be deployed in minutes; a monolith may take hours |

**Disadvantages of Microservices (trade-offs to acknowledge):**

| Trade-off | Description |
|-----------|-------------|
| **Operational complexity** | More services = more moving parts; requires mature DevOps, monitoring, and distributed tracing |
| **Network latency** | Service-to-service calls over a network are slower than in-process calls within a monolith |
| **Distributed data management** | Each service owns its own database; cross-service queries require event-driven patterns, not simple joins |
| **Debugging complexity** | Tracing a bug across multiple services requires distributed tracing tools (e.g., DataDog APM, Jaeger) |

The result was zero-downtime deployments, improved scalability, and 300+ new dealerships onboarded in Year 1 — a concrete, quantified proof point of the advantages listed above.

### 4C. Common Mistakes People Make When Writing Outreach Emails for Technical Roles
