## 2. ⚠️ Key Challenges
---

### 2.1 Operational Challenges

- **Distributed Teams:** Engineering in Bangalore, Product in London, Client stakeholders in New York → timezone overlap of only ~2 hours/day
- **Legacy Systems:** Client has 12+ data sources in 4 different formats (CSV, Oracle DB, REST APIs, Kafka streams)
- **Governance & Compliance:** PII data across pipelines requires GDPR and SOC 2 compliance from day one
- **Change Management:** Business stakeholders resistant to migrating away from Excel-based reporting

### 2.2 Technical Challenges

- **Data Quality:** Inconsistent schemas, missing values, and duplicate records across source systems
- **Scalability:** Platform must handle 500GB/day ingestion with sub-2-second dashboard query latency
- **ML Model Reliability:** Prediction models (sales forecasting, churn) need >85% accuracy before production sign-off
- **Integration Complexity:** 8 downstream systems (ERP, CRM, BI tools) need API contracts defined and maintained

### 2.3 Delivery Challenges

- **Scope Creep Risk:** 6 business units each with unique analytics requirements → risk of expanding scope mid-delivery
- **Resource Constraints:** 3 senior engineers shared across two concurrent client programmes
- **Timeline Pressure:** Go-live deadline tied to client's annual board presentation (non-negotiable)
- **Vendor Dependencies:** Cloud infrastructure provisioning dependent on client's internal IT approval (historically 6–8 week lead time)

### 2.4 Risk & Constraint Summary

| Risk | Likelihood | Impact | Priority |
|------|-----------|--------|----------|
| Scope creep from business units | High | High | 🔴 Critical |
| Data quality issues blocking ML | High | High | 🔴 Critical |

| Key engineer attrition | Low | High | 🟠 High |

## 3. 💡 Proposed Solution
---

### 3.1 Product Thinking

The platform will be treated as an **internal product**, not a one-time project. This means:
- Defining a **product vision**: *'A single source of truth for all business analytics, accessible to every stakeholder in under 5 clicks'*
- Publishing a **public product roadmap** shared with all business unit heads quarterly
- Creating a **feedback loop** with end-users via fortnightly office hours and a shared Slack channel

### 3.2 Process Improvements

| Current State | Future State |
|--------------|-------------|
| 12 siloed data pipelines, manual ETL | Unified data lakehouse with automated ELT |
| Weekly Excel reports emailed to 200+ people | Self-serve BI dashboards with real-time data |
| Ad-hoc ML experiments in local Jupyter notebooks | MLOps pipeline with CI/CD for model deployment |
| Data quality checked manually | Automated data quality framework (Great Expectations) |
| No centralised data catalogue | Data catalogue with lineage (Apache Atlas / Datahub) |

### 3.3 Delivery Strategy

**Approach:** Agile delivery with 2-week sprints, structured around three major platform layers:

1. **Foundation Layer (Months 1–3):** Data ingestion, data lake, governance, security
2. **Insight Layer (Months 3–6):** Data transformation, data warehouse, BI dashboards
3. **Intelligence Layer (Months 6–9):** ML models, forecasting, automated alerting

**Key Principles:**
- *Value First:* Each sprint delivers working, demonstrable functionality
- *Fail Fast:* PoCs validated in Weeks 1–3 before full build begins
- *Freeze Scope:* Hard scope freeze after Sprint 6 (Month 3)
- *Automate Everything:* CI/CD, testing, monitoring from Sprint 1

## 4. 🏗️ System / Process Design

### 4.1 High-Level Architecture

```
┌─────────────────────────────────────────────────────────────────────────┐
│                        DATA SOURCES (Layer 0)                          │
│   Oracle DB │ REST APIs │ Kafka Streams │ CSV/Excel │ CRM │ ERP       │

│              Data Catalogue │ Data Lineage │ PII Masking              │
└────────────────────────────┬────────────────────────────────────────────┘
                             │  (ELT via dbt)
┌────────────────────────────▼────────────────────────────────────────────┐
│                   DATA WAREHOUSE (Layer 2)                             │
│        Snowflake / BigQuery — Dimensional model (Star Schema)         │
│           Automated data quality checks │ SLA monitoring              │
└──────────────┬──────────────────────────┬──────────────────────────────┘
               │                          │
┌──────────────▼───────────┐  ┌───────────▼──────────────────────────────┐

│   Tableau / Looker /     │  │   Sales Forecast │ Churn Model          │
│   Power BI               │  │   Automated retraining │ A/B Testing    │
│   Self-serve analytics   │  │   Model monitoring & drift detection     │
└──────────────────────────┘  └──────────────────────────────────────────┘
```

### 4.2 Integration Points

| Integration | Protocol | Frequency | Owner |

| Data Catalogue → All Layers | Apache Atlas API | Continuous | Platform |

### 4.3 Data Flow

```
Source → Ingest (Kafka/Airflow) → Land (Raw Lake) → 
Quality Check → Transform (dbt) → Serve (DWH) → 
Visualise (BI) / Model (ML) → Alert/Action
```

### 4.4 Automation Opportunities
